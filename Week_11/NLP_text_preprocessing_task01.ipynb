## Libraries and settings

```python
# Libraries
import os
import re
import string
import numpy as np
import pandas as pd
from pprint import pprint

import nltk

# Import only once
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

from nltk.tag import pos_tag
from nltk.corpus import stopwords
from nltk.chunk import tree2conlltags
from nltk.chunk import conlltags2tree
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Current working directory
print('Current working directory:', os.getcwd())
```

## Defining documents

```python
# Original documents (for reference)
# d1 = 'The car is driven on the road.'
# d2 = 'The truck is driven on the highway.'
# d3 = 'The bicycle is driven on the bicycle path.'

# Task 1b: Defining my own documents
d1 = 'The laptop is used for programming and data analysis.'
d2 = 'The smartphone is used for communication and social media.'
d3 = 'The tablet is used for reading and watching videos.'

corpus_01 = d1 + ' ' + d2 + ' ' + d3
corpus_01
```

## Text preprocessing
#### Steps:
- Text to lowercase
- Removing punctuations
- Tokenization
- Removal of stop words
- Lemmatization

### Text to lowercase

```python
# Text to lowercase function
def text_lowercase(text):
    return text.lower()

# Text to lowercase
corpus_02 = text_lowercase(corpus_01)
corpus_02
```

### Removing punctuation

```python
# Remove punctuation function
def remove_punctuation(text):
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)

# Remove punctuation
corpus_03 = remove_punctuation(corpus_02)
corpus_03
```

### Tokenize text & removal of stopwords

```python
# Show english stopwords
eng_stopwords = set(stopwords.words('english'))
print("List of english stopwords:")
print(eng_stopwords)
```

```python
# Function for tokenization and the removal of stopwords
def remove_stopwords(text):
    stop_words = set(stopwords.words("english"))
    word_tokens = word_tokenize(text)
    filtered_text = [word for word in word_tokens if word not in stop_words]
    return filtered_text
 
# Remove stopwords
corpus_04 = remove_stopwords(corpus_03)
print(corpus_04, end="")
```

### Lemmatization

```python
# Initialize Lemmatizer
lemmatizer = WordNetLemmatizer()

# Lemmatize string function
def lemmatize_word(text):
    word_tokens = word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(word, pos ='v') for word in word_tokens]
    return lemmas

# Lemmatize
lem = []
for i in corpus_04:
    lem.append(lemmatize_word(i))

# Nested list to list
corpus_05 = [' '.join([str(x) for x in lst]) for lst in lem]

print('Before lemmatization:')
print(corpus_04, '\n')

print('After lemmatization:')
print(corpus_05, end="")
```

## Redefine the text corpus (pre-processed)

```python
# Original corpus (for reference)
# corpus = ['car drive road', 
#           'truck drive highway', 
#           'bicycle drive bicycle path']

# Task 1c: Redefine corpus with my pre-processed and lemmatized words
# After removing stopwords, punctuation, and lemmatization, my documents become:
# d1: laptop, use, program, data, analysis
# d2: smartphone, use, communication, social, media
# d3: tablet, use, read, watch, video

corpus = ['laptop use program data analysis', 
          'smartphone use communication social media', 
          'tablet use read watch video']
```

## Task 1d: Document-term matrix with ngram_range=(1,1)

```python
# Vectorizer with ngram_range=(1,1)
vectorizer = CountVectorizer(min_df=0.0, ngram_range=(1,1))

# Transform 
count = vectorizer.fit_transform(corpus)
 
# Create dataframe
df_count = pd.DataFrame(count.toarray(),
                        columns=vectorizer.get_feature_names_out())

print('Document-term matrix')
print(df_count)
```

## Task 1d: Document-term matrix with ngram_range=(2,2)

```python
# Vectorizer with with ngram_range=(2,2)
vectorizer = CountVectorizer(min_df=0.0, ngram_range=(2,2))

# Transform 
count = vectorizer.fit_transform(corpus)
 
# Create dataframe
df_count = pd.DataFrame(count.toarray(),
                        columns=vectorizer.get_feature_names_out())

print('Document-term matrix')
print(df_count)
```

## Task 1e: Term frequency-inverse document frequency (TF-IDF)
- For details see: https://www.learndatasci.com/glossary/tf-idf-term-frequency-inverse-document-frequency

### Term Frequency (TF)

```python
# Compute Term Frequency (TF)
words_set = set()
for doc in corpus:
    words = doc.split(' ')
    words_set = words_set.union(set(words))
    
print('Number of words in the corpus:',len(words_set), '\n')
print('The words in the corpus: \n', words_set)

# Number of documents in the corpus
n_docs = len(corpus)

# Number of unique words in the corpus 
n_words_set = len(words_set)

df_tf = pd.DataFrame(np.zeros((n_docs, n_words_set)), 
                     columns=list(words_set))

print("\nTerm Frequency (TF):")
for i in range(n_docs):
    # Words in the document
    words = corpus[i].split(' ')
    for w in words:
        df_tf[w][i] = df_tf[w][i] + (1 / len(words))
        
print(df_tf.round(4))
```

### Inverse Document Frequency (IDF)

```python
# Computing Inverse Document Frequency (IDF)
print("\nInverse Document Frequency (IDF):")

idf = {}

for w in words_set:
    
    # k = number of documents that contain this word
    k = 0
    
    for i in range(n_docs):
        if w in corpus[i].split():
            k += 1
            
    idf[w] =  np.log10(n_docs / k).round(4)
    
    print(f'{w:>15}: {idf[w]:>10}')
```

### Term Frequency - Inverse Document Frequency (TF-IDF)

```python
# Computing TF-IDF
df_tf_idf = df_tf.copy()

for w in words_set:
    for i in range(n_docs):
        df_tf_idf[w][i] = df_tf[w][i] * idf[w]

print('\nTF-IDF:')
print(df_tf_idf.round(4))
```

## Task 1f: Part-of-Speech (POS) tagging
For meaning of POS-tags see: https://pythonexamples.org/nltk-pos-tagging

```python
# Task 1f: My own text example for POS tagging
text = '''Artificial intelligence is transforming the technology industry. 
          Companies are investing heavily in machine learning and data science 
          to develop innovative solutions.'''

def preprocess(sent):
    sent = nltk.word_tokenize(sent)
    sent = nltk.pos_tag(sent)
    return sent

sent = preprocess(text)
pattern = 'NP: {<DT>?<JJ>*<NN>}'

cp = nltk.RegexpParser(pattern)
cs = cp.parse(sent)

iob_tagged = tree2conlltags(cs)

# Print the POS-tags
pprint(iob_tagged)
```

### Explanation of POS Tags from the Output

Based on the POS tagging output above, here are the explanations of at least 5 common POS tags:

1. **NN (Noun, singular)**: 
   - Represents a singular noun (person, place, thing, or idea)
   - Examples from my text: "intelligence", "industry", "learning", "science"
   - These are single entities or concepts

2. **NNS (Noun, plural)**: 
   - Represents a plural noun
   - Examples from my text: "Companies", "solutions"
   - Indicates multiple instances of a noun

3. **JJ (Adjective)**: 
   - Describes or modifies a noun
   - Examples from my text: "Artificial", "innovative"
   - Provides additional information about the characteristics of nouns

4. **VBG (Verb, gerund or present participle)**: 
   - A verb form ending in '-ing' that functions as a noun or adjective
   - Examples from my text: "transforming", "investing"
   - Can indicate ongoing actions or processes

5. **VBZ (Verb, 3rd person singular present)**: 
   - A verb in present tense for third person singular (he/she/it)
   - Example from my text: "is"
   - Shows the relationship or state of being

6. **DT (Determiner)**: 
   - Words that introduce nouns and specify their reference
   - Example from my text: "the"
   - Helps identify whether the noun is specific or general

7. **IN (Preposition or subordinating conjunction)**: 
   - Shows relationships between words in terms of time, place, direction, etc.
   - Examples from my text: "in", "to"
   - Connects nouns or pronouns to other words in the sentence
